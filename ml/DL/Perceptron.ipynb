```python
import torch
import torch.nn as nn 
```

```python
# input 
torch.manual_seed(42) # each time we 
inputs= torch.tensor([1.0,2.0,3.0]) # tensor here can be used to create multidimentional array.
```

```python
neuron= nn.Linear(in_features=3, out_features=1) # here we are creating a neuron with "linear" activation function with 3 i/p and 1 o/p.
```

```python
output= neuron(inputs) # getting outputs 
```

```python
output # since w1, w2 ,w3 ,b are randomly initialised so the ans. is also random
```

```python
# verifying by fromula
neuron.weight @ inputs + neuron.bias
```
