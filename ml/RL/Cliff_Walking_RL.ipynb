# SARSA

```python
# Defining parameters  
gamma=0.99 
alpha= 0.5 # learning rate
epsilon= 0.1 #10% exploring, 90% exploiting
episodes= 500
```

```python
# Q-table => For storing Q-values 
Q= np.zeros((48,4))
```

```python
# Policy- E-greedy=> state --> action  
def epsilon_greedy(state): 
    if random.random() < epsilon: 
        return env.action_space.sample() # This will choose a random action to explore from action space.
    else: 
        return np.argmax(Q[state]) # This will return the max q-value index for that state to exploit, means best action for that state. 
```

```python
env= gym.make("CliffWalking-v1") # Creating ones and updating again
for episode in range(episodes): 
    done= False # A pameter stating episode ended 
    state,_ = env.reset() 
    action= epsilon_greedy(state) 

    total_reward= 0 # For each episode 
    episode_len= 0 # How many movements in each episode 

    while not done: 
        next_state, reward,terminated, truncated,_= env.step(action) # it gives 5 values
        done= terminated or truncated
        if done:
            Q[state, action] += alpha*(reward- Q[state, action]) # if done then there is no future state so updation on current reward only.
        else:
            next_action= epsilon_greedy(next_state)

        # SARSA update 
            Q[state, action] += alpha*(reward+ gamma*Q[next_state, next_action] - Q[state, action]) 

            state= next_state 
            action= next_action 

        total_reward+= reward 
        episode_len+= 1 

    print(f"episode= {episode+1}/500: total reward ={total_reward} & ep length= {episode_len}")
env.close()
```

```python
# Visualising the learning
env= gym.make("CliffWalking-v1", render_mode= "human")
state,_= env.reset()
done= False 
total_reward=0 
episode_len= 0 

while not done: 
    action= np.argmax(Q[state])
    state, reward, terminated, truncated,_= env.step(action) # Here we have named next_state as state so loop can take the updated value without any extra line. 
    done= terminated or truncated 

    total_reward+= reward 
    episode_len+= 1  

print(f"total reward={total_reward}, episode length={episode_len}")
env.close()
```

# Q-learning

```python
# Using the same parameters,E-policy & q-table 

env= gym.make("CliffWalking-v1")
for episode in range(episodes):
    done= False 
    state,_= env.reset() 
    action= epsilon_greedy(state) 
    total_reward=0 
    episode_len=0 

    while not done: 
        next_state, reward, terminated, truncated, _=env.step(action) 
        done= terminated or truncated 

        if done: 
            Q[state, action] += alpha*(reward- Q[state, action])
        else: 
            Q[state, action]+= alpha*(reward+ gamma* np.max(Q[next_state])- Q[state, action]) # np max will give max value(action) for that state

            state= next_state 
            action= epsilon_greedy(state) 

        total_reward+= reward 
        episode_len+= 1 

    print(f"epidode={episode+1}/500 ,total reward={total_reward} & episode len.={episode_len}")

env.close()
```
