```python
from sklearn.datasets import load_iris
import seaborn as sns 
from sklearn.preprocessing import StandardScaler
```

```python
iris= load_iris()

X= iris.data
y= iris.target
```

```python
scaler= StandardScaler()
X_scaled= scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X[:,0], y=X[:,2])
```

```python
# DBSCAN 
from sklearn.cluster import DBSCAN

dbscan= DBSCAN(
    eps=0.8, # random value for know
    min_samples= 5
 )
```

```python
lables= dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X[:,0],y=X[:,2], c=lables)
```

```python
lables # here 0 in cluster 1,cluster 2 is 1 qnd -1 is outlier/noise
# its a linear dataset and DBSCAN is performing poor, since we already know iris has 3 clustures but it sayes 2 only ,that too it automatically findes and executes
```

# DBSCAN on non-linear data

```python
from sklearn.datasets import make_moons # this provides non-linear datasets
```

```python
X,y= make_moons(
    n_samples=300, 
    noise=0.05, 
    random_state=44
)
```

```python
X_scaled= scaler.fit_transform(X)

sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1])
```

```python
from sklearn.cluster import KMeans

kmean= KMeans(
    n_clusters=2,
    random_state=44
)

klabel= kmean.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], c=klabel)
```

```python
dbscan= DBSCAN(
    eps=0.3, 
    min_samples=5
)

db_lables= dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], c=db_lables)
```

# Anomaly Detection 

```python
X,y= make_moons(
    n_samples=500, 
    noise= 0.1,
    random_state=42
)
```

```python
X_scaled= scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1])
```

```python
db= DBSCAN(
    eps= 0.2,
    min_samples=5
)

labels= db.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],c=labels)
```
