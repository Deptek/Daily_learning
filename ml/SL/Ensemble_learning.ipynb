# Boosting using, gradient boosting

```python
# Regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score ,accuracy_score, classification_report
from sklearn.datasets import make_regression # This will help ous to generate our own data 
```

```python
# generating data 
X,y= make_regression(
    n_features=10, 
    n_samples=1000, # actual data points 
    noise=20, 
    random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.3
)
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
gbr= GradientBoostingRegressor(   # most of its parameters are same as decision tree since its using it as default
    n_estimators=200, # total 200 bse decision trees
    learning_rate=0.05, 
    random_state=42,
    max_depth=3, 
    subsample=0.8 # this tells how much part of data (80%) will be feeded to every base learner
)
```

```python
gbr.fit(X_train,y_train)
```

```python
y_pred= gbr.predict(X_test)
```

```python
print("R2_score:",r2_score(y_test, y_pred))
```

```python
# Classifier
from sklearn.datasets import make_classification # by-default it will generate data for binary classification

X,y= make_classification(
    n_features=20, 
    n_samples=500, # actual data points 
    n_informative=10, # this says only 10 features out of 20 are usefull ,so know model has more noise
    random_state=42
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.3
)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gbc= GradientBoostingClassifier(   
    n_estimators=150, 
    learning_rate=0.1, 
    random_state=42,
    max_depth=3 
)
```

```python
gbc.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred=gbc.predict(X_test)
print("R2_score:",r2_score(y_test, y_pred))
print("Accuracy_score:",accuracy_score(y_test, y_pred))
```

# Ada boost

```python
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
```

```python
X,y= make_classification(
    n_samples=500, 
    n_features=20, 
    n_informative=10, 
    n_redundant=2, # means 2 features will have repeating data
    random_state=42
)

X_train,X_test,y_train,y_test= train_test_split(X,y,random_state=42, test_size=0.3)
```

```python
from sklearn.ensemble import AdaBoostClassifier

# DT stump
base_model= DecisionTreeClassifier(
    max_depth=1
)

#adaboost classifier 
abc= AdaBoostClassifier(  # if we dont give an estimator it by default take a stump (DT with max_depth =1)
    estimator=base_model, 
    n_estimators=100, 
    random_state=42,
    algorithm="SAMME" # this is one of the boosting stategy to combine thoe weak learners ,this one is future proof.
)
```

```python
abc.fit(X_train,y_train)
```

```python
y_pred= abc.predict(X_test)
```

```python
print("Accuracy_score:",accuracy_score(y_test, y_pred))
print("Classification_report:",classification_report(y_test, y_pred))

# similarly we can generate a regressor.
```

```python
import xgboost as xgb 
from sklearn.datasets import make_classification
```

```python
X,y= make_classification(
    n_features=20, 
    n_samples= 100, 
    random_state=42, 
    n_redundant=10, 
    n_informative=10
)

X_train,X_test,y_train,y_test= train_test_split(
    X,y, test_size=0.3, random_state=42
)
```

```python
xgb_class =xgb.XGBClassifier(
    n_estimators=100, 
    max_depth=3, 
    learning_rate=0.1, 
    eval_metric="logloss",
    random_state= 42  # evaluation metric
)
```

```python
xgb_class.fit(X_train,y_train)
```

```python
y_pred=xgb_class.predict(X_test)

print("Accur.:",accuracy_score(y_test,y_pred))
print("Clasi.report:",classification_report(y_test,y_pred))

# Similarly xg boost reg. can be applied
```
